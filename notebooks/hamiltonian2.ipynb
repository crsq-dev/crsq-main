```python
from crsq.blocks import (hamiltonian2, wave_function)

num_coord_qubits = 4
wfr = wave_function.WaveFunctionRegisterSpec(1, num_coord_qubits, 16, 1, 0, 1)
ham2 = hamiltonian2.InverseSquareRoot(wfr)

print("circuit qubit count: ", ham2.circuit.num_qubits)
ham2.circuit.draw(scale=0.4, fold=80)

```
