```python
import matplotlib.pyplot as plt
import numpy
import crsq.models.hydrogen3d as hydrogen3d
import math

psi3d = hydrogen3d.PsiH3D(0,0,0,1,0,0)

L = 10
M = 512
dq = L/M
q = numpy.linspace(-L/2, L/2-dq, M)
x,y,z=numpy.meshgrid(q,q,q)
psi = psi3d(x,y,z)*math.pow(dq,3/2)
psidagger = numpy.conjugate(psi)

sum = numpy.sum(psidagger*psi)
print(sum)
```
